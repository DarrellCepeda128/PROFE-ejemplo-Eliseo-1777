```python
import warnings
warnings.filterwarnings('ignore')

# Clase automovil, atributos, marca, placas, precio
# funciones leerdatos(), mostrardatos()
# crear 3 objetos, carro1, carro2, carro3

class Automovil:
    def __init__(self, marca, placas, precio):
        self.marca = marca
        self.placas = placas
        self.precio = precio

    def leerdatos(self):
        self.marca = input("Ingrese la marca del automovil: ")
        self.placas = input("Ingrese las placas del automovil: ")
        self.precio = float(input("Ingrese el precio del automovil: "))
        # The method updates self's attributes, no need to return and reassign to self.
        # If a return value is desired, it should be handled by the caller, not reassigned to the object itself.
        return self.marca, self.placas, self.precio

    def mostrardatos(self):
        print("Marca:", self.marca)
        print("Placas:", self.placas) # Corrected typo: self.placa to self.placas
        print("Precio:", self.precio)
        # The method's primary purpose is to print, return value can be ignored if not used.
        return self.marca, self.placas, self.precio
# fin de la clase automovil

# Instantiate each Automovil object individually
carro1 = Automovil("", "", 0.0)
carro2 = Automovil("", "", 0.0)
carro3 = Automovil("", "", 0.0)

print("Carro Numero 1")
carro1.leerdatos()
carro1.mostrardatos()
print("-+-+-+-+-+-+-+-+-+")

print("Carro Numero 2")
carro2.leerdatos()
carro2.mostrardatos()
print("-+-+-+-+-+-+-+-+-+")

print("Carro Numero 3")
carro3.leerdatos()
carro3.mostrardatos()
print("-+-+-+-+-+-+-+-+-+")

print("software realizado por Cepeda Darrell NL = 10")
```
